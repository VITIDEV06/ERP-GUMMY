```python
from pathlib import Path

# Configuración
INPUT_DIR = Path(".")
OUTPUT_DIR = Path("CLEAN")
MIN_FREQUENCY = 7

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Cargar los datasets
orders = pd.read_csv(INPUT_DIR / "olist_orders_dataset.csv")
items = pd.read_csv(INPUT_DIR / "olist_order_items_dataset.csv")
payments = pd.read_csv(INPUT_DIR / "olist_order_payments_dataset.csv")
customers = pd.read_csv(INPUT_DIR / "olist_customers_dataset.csv")
products = pd.read_csv(INPUT_DIR / "olist_products_dataset.csv")
sellers = pd.read_csv(INPUT_DIR / "olist_sellers_dataset.csv")


# ==========================================
# 1. Filtrar pedidos según su frecuencia
# ==========================================

frequency = items["order_id"].value_counts()

valid_order_ids = frequency[frequency > MIN_FREQUENCY].index

filtered_orders = orders.loc[
    orders["order_id"].isin(valid_order_ids)
].copy()

# Agregar la frecuencia correspondiente a cada pedido
filtered_orders["frequency_in_items"] = (
    filtered_orders["order_id"].map(frequency)
)


# ==========================================
# 2. Filtrar items, payments y customers
# ==========================================

filtered_items = items.loc[
    items["order_id"].isin(filtered_orders["order_id"])
].copy()

filtered_payments = payments.loc[
    payments["order_id"].isin(filtered_orders["order_id"])
].copy()

filtered_customers = customers.loc[
    customers["customer_id"].isin(filtered_orders["customer_id"])
].copy()


# ==========================================
# 3. Filtrar products
# ==========================================

filtered_products = products.loc[
    products["product_id"].isin(filtered_items["product_id"])
].copy()


# ==========================================
# 4. Filtrar sellers
# ==========================================

filtered_sellers = sellers.loc[
    sellers["seller_id"].isin(filtered_items["seller_id"])
].copy()


# ==========================================
# 5. Exportar los datasets
# ==========================================

datasets = {
    "olist_orders_dataset.csv": filtered_orders,
    "olist_order_items_dataset.csv": filtered_items,
    "olist_order_payments_dataset.csv": filtered_payments,
    "olist_customers_dataset.csv": filtered_customers,
    "olist_products_dataset.csv": filtered_products,
    "olist_sellers_dataset.csv": filtered_sellers
}

for filename, dataframe in datasets.items():
    dataframe.to_csv(OUTPUT_DIR / filename, index=False)

    print(f"{filename}: {len(dataframe)} filas conservadas")


print("\nPipeline ejecutado correctamente.")
```
